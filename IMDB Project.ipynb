```python
df[['Series_Title','IMDB_Rating']].sort_values(by='IMDB_Rating', ascending=False).head(10)
```

```python
plt.figure(figsize=(10,5))
top10 = df[['Series_Title','IMDB_Rating']].sort_values(by='IMDB_Rating', ascending=False).head(10)
sns.barplot(x='IMDB_Rating', y='Series_Title', data=top10)
plt.title("Top 10 Movies by IMDb Rating")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['IMDB_Rating'], bins=20, kde=True)
plt.title("Distribution of IMDb Ratings")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['IMDB_Rating'])
plt.title("Box Plot of IMDb Ratings")
plt.show()
```

```python
df['Genre'].value_counts().head(10)
```

```python
plt.figure(figsize=(10,5))
df['Genre'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.show()
```

```python
df[['IMDB_Rating','Meta_score']].corr()
```

```python
plt.figure(figsize=(5,4))
sns.heatmap(df[['IMDB_Rating','Meta_score']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Meta_score', y='IMDB_Rating', data=df)
plt.title("Meta Score vs IMDb Rating")
plt.show()
```

```python
df['Director'].value_counts().head(10)
```
